# Scan

Scanning allows inspecting large surveys without loading everything in
memory.

```python
import pysegy as seg

# scan all matching files in the repository data directory
data_dir = "../data"
scan = seg.segy_scan(data_dir, "overthrust_2D_shot_*.segy")

print(len(scan.shots))  # number of shot records
print(scan.counts[0])   # traces in the first shot
```

The [pysegy.segy_scan](reference/segy_scan.html#pysegy.segy_scan)
function summarises multiple files and returns a
[pysegy.SegyScan](reference/SegyScan.html#pysegy.SegyScan) that can
lazily read data for each shot. Use `by_receiver=True` to group traces
by receiver coordinates instead of source coordinates.

Depth field detection - The source and receiver depth fields are
auto-detected using a fixed priority, not by picking headers with the
largest value spread. This reduces surprises with unlikely headers. -
Source priority: `SourceDepth` \> `SourceDatumElevation` \>
`SourceSurfaceElevation` \> `SourceWaterDepth`. - Receiver priority:
`RecGroupElevation` \> `RecDatumElevation` \> `GroupWaterDepth`.
