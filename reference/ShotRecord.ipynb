# ShotRecord

``` python
ShotRecord(
    path,
    coordinates,
    fileheader,
    rec_depth_key='RecGroupElevation',
    depth_key='SourceDepth',
    by_receiver=False,
    segments=list(),
    summary=dict(),
    ns=0,
    dt=0,
    fs=None,
)
```

Information about a single shot or receiver gather within a SEGY file.

## Attributes

| Name | Description |
|------------------------------------|------------------------------------|
| [by_receiver](#pysegy.ShotRecord.by_receiver) | bool(x) -\> bool |
| [data](#pysegy.ShotRecord.data) | The traces of this record, read when they are indexed. |
| [depth_key](#pysegy.ShotRecord.depth_key) | str(object=’’) -\> str |
| [dt](#pysegy.ShotRecord.dt) | int(\[x\]) -\> integer |
| [ns](#pysegy.ShotRecord.ns) | int(\[x\]) -\> integer |
| [ntraces](#pysegy.ShotRecord.ntraces) | Number of traces in this record. |
| [rec_coordinates](#pysegy.ShotRecord.rec_coordinates) | Array of receiver coordinates for this gather. |
| [rec_depth_key](#pysegy.ShotRecord.rec_depth_key) | str(object=’’) -\> str |
| [traceheaders](#pysegy.ShotRecord.traceheaders) | Load trace headers for all shots on first access. |

## Methods

| Name | Description |
|------------------------------------|------------------------------------|
| [read_data](#pysegy.ShotRecord.read_data) | Load the traces of this shot. |
| [read_header_fields](#pysegy.ShotRecord.read_header_fields) | Values of `keys` for every trace of this shot, as scaled arrays. |
| [read_headers](#pysegy.ShotRecord.read_headers) | Read only the headers for this shot. |

### read_data

``` python
read_data(keys=None, traces=None)
```

Load the traces of this shot.

#### Parameters

| Name | Type | Description | Default |
|-----|---------|-----------------------------------------------------|------|
| keys | Iterable\[str\] | Header fields to load with each trace. | `None` |
| traces | slice | Range of this record’s traces to load. Only the corresponding part of the file is read; by default all traces are. | `None` |

#### Returns

| Name | Type    | Description                   |
|------|---------|-------------------------------|
|      | ndarray | `ns` x number of traces read. |

### read_header_fields

``` python
read_header_fields(keys)
```

Values of `keys` for every trace of this shot, as scaled arrays.

The array form of :meth:`read_headers`, decoding a field at a time
across the block the way a scan does, rather than building a header
object per trace.

#### Parameters

| Name | Type            | Description             | Default    |
|------|-----------------|-------------------------|------------|
| keys | Iterable\[str\] | Header words to return. | *required* |

#### Returns

| Name | Type | Description                                       |
|------|------|---------------------------------------------------|
|      | dict | `{key: ndarray}`, one scaled float array per key. |

### read_headers

``` python
read_headers(keys=None)
```

Read only the headers for this shot.
